# Market Performance & Portfolio Analytics

Exploratory analysis of the MarketScope equity universe: performance KPIs,
sector trends, correlations and top movers. The same functions power the
Streamlit dashboard in `app/dashboard.py`.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt
import seaborn as sns

from src.make_dataset import load_market_data
from src.analysis import (
    compute_kpis, sector_index, market_index,
    correlation_matrix, top_movers, headline,
)

sns.set_theme(style="whitegrid")
%matplotlib inline
```

```python
prices = load_market_data()
print(prices.shape)
prices.head()
```

## Headline market stats

```python
headline(prices)
```

## Per-ticker KPIs

```python
kpis = compute_kpis(prices)
kpis.head(10)
```

## Sector performance (equal-weight, rebased to 1.0)

```python
curve = sector_index(prices)
curve.plot(figsize=(10, 5), title="Cumulative sector performance")
plt.ylabel("Index (base = 1.0)")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.show()
```

## Correlation of daily returns (sector level)

```python
corr = correlation_matrix(prices, level="sector")
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", center=0, vmin=-1, vmax=1, ax=ax)
ax.set_title("Sector daily-return correlation")
plt.show()
```

```python
top_movers(prices, window=30)
```
